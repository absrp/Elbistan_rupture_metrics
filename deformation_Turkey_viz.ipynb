```python
import pandas as pd 
import matplotlib.pyplot as plt 
import geopandas as gpd 
import numpy as np 
```

### Import data from Liu et al. 
Data link: https://springernature.figshare.com/articles/dataset/The_post-processed_data_presented_in_the_paper_of_Extensive_off-fault_damage_around_the_2023_Kahramanmara_T_rkiye_earthquake_surface_ruptures_/25801234?file=49680690

```python
aftershock_OFD_data = pd.read_csv('Turkey_data_Liu/TotalOnFaultDisp_Mw7.6.txt', sep='\t', 
                                 names=['Lon(deg)', 'Lat(deg)', 'DistanceFromSW(km)', 'Total(m)', 'On-fault(m)'],
                                 usecols=[0,1,2,3,4],skiprows=1)
```

```python
fig, axes = plt.subplots(2, 3,dpi=300,figsize=(10,6))
axes[0,0].scatter(aftershock_OFD_data['Lon(deg)'], aftershock_OFD_data['Lat(deg)'], c=aftershock_OFD_data['Total(m)'], s=5,cmap='magma')
axes[0,0].set_title('Total deformation')
axes[0,0].set_xlabel('Longitude')
axes[0,0].set_ylabel('Latitude')
axes[0,0].set_aspect('equal')

axes[0,1].scatter(aftershock_OFD_data['Lon(deg)'], aftershock_OFD_data['Lat(deg)'], c=aftershock_OFD_data['On-fault(m)'],s=5, cmap='magma')
axes[0,1].set_title('On-fault deformation')
axes[0,1].set_xlabel('Longitude')
axes[0,1].set_ylabel('Latitude')
axes[0,1].set_aspect('equal')

axes[0,2].scatter(aftershock_OFD_data['Lon(deg)'], aftershock_OFD_data['Lat(deg)'], c=aftershock_OFD_data['Total(m)']-aftershock_OFD_data['On-fault(m)'],s=5, cmap='magma')
axes[0,2].set_title('Off-fault deformation')
axes[0,2].set_xlabel('Longitude')
axes[0,2].set_ylabel('Latitude')
axes[0,2].set_aspect('equal')

axes[1,0].scatter(aftershock_OFD_data['DistanceFromSW(km)'], aftershock_OFD_data['Total(m)'],s=8,c='darkorange')
axes[1,0].set_xlabel('Distance along strike (km)')
axes[1,0].set_ylabel('Total deformation (m)')

axes[1,1].scatter(aftershock_OFD_data['DistanceFromSW(km)'], aftershock_OFD_data['On-fault(m)'],s=8,c='darkorange')
axes[1,1].set_xlabel('Distance along strike (km)')
axes[1,1].set_ylabel('On-fault deformation (m)')

axes[1,2].scatter(aftershock_OFD_data['DistanceFromSW(km)'], aftershock_OFD_data['Total(m)']-aftershock_OFD_data['On-fault(m)'],s=10,c='darkorange')
axes[1,2].set_xlabel('Distance along strike (km)')
axes[1,2].set_ylabel('Off-fault deformation (m)')

plt.tight_layout()
plt.savefig('deformation_Turkey_aftershock_viz.png',dpi=300,bbox_inches='tight')
```
